```python
from pytanga.geometry import Geometry, Point
from pytanga.basis import BasisPGA3
from pytanga.viz import Visualizer
import math

P3 = BasisPGA3()
geo = Geometry(P3)
viz = Visualizer()
# viz.start_server()
```

```python
with viz:
    a = geo(Point(1, 0, 0))
    b = geo(Point(0, 1, 0))
    viz.add(a, color="orange")
    viz.add(b, color="blue")

    time = 0.0
    for dt in viz.animate(fps=30, auto_clear=True):
        time += dt
        y = -1.0 + 2 * math.sin(time * math.pi)
        c = geo(Point(0.5, y, y))
        viz.add(c, color="green")

        line = b.join(a)
        # line.show("line")
        viz.add(line, color="green")

        # c.show("c")
        plane = (c.dual() ^ line.dual()).dual()
        # plane.show("plane")
        # print(geo(plane))
        viz.add(plane)
```
