```python
from pytanga.geometry import Geometry, Point
from pytanga.basis import BasisPGA3
from pytanga.viz import Visualizer

P3 = BasisPGA3()
geo = Geometry(P3)
viz = Visualizer()
# viz.start_server()
```

```python
with viz:
    a = geo(Point(1, 0, 0))
    b = geo(Point(0, 1, 0))
    c = geo(Point(0.5, 0.5, 0))
    viz.add(a, color="orange")
    viz.add(b, color="blue")
    viz.add(c, color="green")

    line = b.join(a)
    line.show("line")
    viz.add(line, color="green")

    c.show("c")
    plane = c.join(line)
    plane.show("plane")
```
