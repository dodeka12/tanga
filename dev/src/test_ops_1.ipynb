```python
from pytanga.geometry import Geometry, Point, Translator, Direction
from pytanga.basis import BasisPGA3

B3 = BasisPGA3()
geo = Geometry(B3)
```

```python
p = geo.create(Point(1,0,0))

# reflection in plane normal to e1 passing through origin
op = B3("e1")
q = op.vp(p)
q.show("q")
print(geo.which_entity(q))
```

```python
p = geo.create(Point(0,1,0))

# this operator maps everything to zero when applied as versor product.
op = B3.e1 ^ B3.e0
q = op.vp(p)
q.show("q")
# print(geo.which_entity(q))
```

```python
p = geo.create(Point(0,1,0))

# this operator maps everything to zero when applied as versor product.
op = geo.create(Point(1,0,0))
q = op.vp(p)
q.show("q")
print(geo.which_entity(q))
```
