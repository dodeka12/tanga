```python
from pytanga.geometry import Geometry, Point, Translator, Direction, Sphere, PointPair
from pytanga.basis import BasisPGA3, BasisN2, BasisN3

B3 = BasisPGA3()
N3 = BasisN3()
N2 = BasisN2()

```

```python

geo = Geometry(N3)
sphere = geo(Sphere(center=Point(0, 0, 0), radius=2))
sphere.show("sphere")
ana = geo(sphere)
print(ana)

A = geo(Point(-1,0,0))
B = geo(Point(1,0,0))
C = geo(Point(0,1,0))
D = geo(Point(0,0,1))
A.show("A")
sphere2 = A^B^C^D
sphere2.show("sphere2")
ana2 = geo(sphere2)
print(ana2)
```

```python

geo = Geometry(N2)
A = geo(Point(-1,0))
B = geo(Point(1,0))
C = geo(Point(0,1))

A.show("A")
B.show("B")
C.show("C")

circle = A^B^C
circle.show("circle")

ana = geo.analyze(circle)
print(ana)

c_imag = geo(Point(0,0)) + 0.5*N2.einf
c_imag.show("c_imag")
ana = geo(c_imag.dual())
print(ana)
print(type(ana))

```

```python
geo = Geometry(N2)
pp = geo(PointPair(Point(-1,-1), Point(1,1)))
pp.show("pp")
pp = 1.0 * pp
pp.show("scaled pp")

L = pp ^ N2.einf
L.show("L")
print("Line from pp:")
print(geo(L))

P_star = pp | N2.einf
P_star.show("P_star")
print("Bisector plane from pp:")
print(geo(P_star.dual()))

# direction from line
dir = L | N2.E
dir.show("dir")
print("Direction from line:")
print(geo(dir))

X = P_star | L
X.show("X")
print("Intersection point from line and plane:")
print(geo(X))

L_inv = L.inv()
L_inv.show("L_inv")

S_star = pp | L_inv
S_star.show("S_star")
r_sq = (S_star | S_star).scalar
print(f"Radius squared from pp: {r_sq}")

print(geo(pp))
```
