```python
from pytanga.geometry import Geometry, Point, Translator, Direction, Sphere
from pytanga.basis import BasisPGA3, BasisN2, BasisN3

B3 = BasisPGA3()
N3 = BasisN3()
N2 = BasisN2()

```

```python

geo = Geometry(N3)
sphere = geo(Sphere(center=Point(0, 0, 0), radius=2))
sphere.show("sphere")
ana = geo(sphere)
print(ana)

A = geo(Point(-1,0,0))
B = geo(Point(1,0,0))
C = geo(Point(0,1,0))
D = geo(Point(0,0,1))
A.show("A")
sphere2 = A^B^C^D
sphere2.show("sphere2")
ana2 = geo(sphere2)
print(ana2)
```

```python

geo = Geometry(N2)
A = geo(Point(-1,0))
B = geo(Point(1,0))
C = geo(Point(0,1))

A.show("A")
B.show("B")
C.show("C")

circle = A^B^C
circle.show("circle")

ana = geo.analyze(circle)
print(ana)

c_imag = geo(Point(0,0)) + 0.5*N2.einf
c_imag.show("c_imag")
ana = geo(c_imag.dual())
print(ana)
print(type(ana))
```
