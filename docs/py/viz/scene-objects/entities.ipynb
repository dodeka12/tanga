# Entities

Every geometric entity you can add to a scene, together with its visualization
style class, the style parameters, and a static inline example. Each example is
rendered serverlessly via `display_snapshot()`.

For operators, see `operators.md`; for axes/grid and point paths, see
`axes-grid.md` and `point-path.md`.


```python
from pytanga.geometry import (
    Circle,
    Direction,
    HPoint,
    ImagCircle,
    ImagPointPair,
    ImagSphere,
    Line,
    Plane,
    Point,
    PointPair,
    Space,
    Sphere,
)
from pytanga.viz import PointStyle, SphereStyle, Visualizer

_NOTE = "You can interactively rotate, pan, and zoom this view."


def show(entity, height=340, **kwargs):
    # Render a single entity in a fresh scene, inline and serverless.
    viz = Visualizer(
        title="Entities - " + type(entity).__name__,
        add_default_axes=False,
        annotation=_NOTE,
    )
    viz.add(entity, **kwargs)
    return viz.display_snapshot(height=height)
```

## Point

A single 3D point. `CrossHairPointStyle` renders a 3D crosshair instead of a
sphere, inheriting `color`, `opacity`, and `size` (plus `arm_thickness`).

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| size | float | Point size in world units |


```python
show(Point(2, 0, 0), color="#ff4444", style=PointStyle(size=0.12))
```

## Direction

An arrow from the origin.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| length | float | Arrow length in world units |


```python
show(Direction(1, 1, 0), color="#ffffff")
```

## HPoint

A homogeneous point.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| size | float | Point size in world units |


```python
show(
    HPoint(point=Point(-2, -2, 1)),
    color="#44ff44",
    style=PointStyle(size=0.12),
)
```

## PointPair

A pair of points connected by a line.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| point_size | float | Size of the endpoint spheres |
| line_thickness | float | Connecting line width (screen pixels) |
| wireframe* | - | Wireframe cage over each point sphere |


```python
show(
    PointPair(point_a=Point(-1, 1, 0), point_b=Point(1, 1, 0)),
    color="#44ff44",
)
```

## Line

An infinite line through the origin. `CylinderLineStyle` renders it as a solid
3D cylinder (`thickness` in world units) instead of a screen-space fat line.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| length | float | Drawn length in world units |
| thickness | float | Line width in screen pixels |
| wireframe* | - | Wireframe cage over the cylinder |


```python
show(
    Line(origin=Point(0, 0, 0), direction=Direction(1, 0, 0)),
    color="#44ff44",
)
```

## Plane

An infinite plane. Supports `texture_label` (see `texture-labels.md`) and
`double_sided`.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| extent | float | Drawn size in world units |
| wireframe* | - | Wireframe cage over the surface |
| texture_label | TextureLabelStyle | Surface text/KaTeX label |
| double_sided | bool | Render both faces |


```python
show(
    Plane(point=Point(0, 0, 2), normal=Direction(0, 0, 1)),
    opacity=0.25,
)
```

## Circle

A circle in a plane.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| wireframe* | - | Wireframe cage over the surface |


```python
show(
    Circle(center=Point(0, 0, 0), normal=Direction(0, 0, 1), radius=2.5),
    color="#ff44ff",
)
```

## Sphere

A sphere. Supports `texture_label` and `double_sided`.

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| wireframe* | - | Wireframe cage over the surface |
| texture_label | TextureLabelStyle | Surface text/KaTeX label |
| double_sided | bool | Render both faces |


```python
show(
    Sphere(Point(0, 0, 0), radius=2),
    style=SphereStyle(wireframe=True),
    opacity=0.3,
)
```

## Space

The entire 3D volume (pseudoscalar).

| Field | Type | Description |
|---|---|---|
| color | str | CSS colour |
| opacity | float | 0-1 |
| extent | float | Bounding-box size in world units |


```python
show(Space(), opacity=0.08)
```

## Imaginary entities

`ImagCircle`, `ImagSphere`, and `ImagPointPair` reuse the style class of their
real counterpart but default to a dotted wireframe (often with a transparent
surface), since they have no real Euclidean points.

| Entity | Style class | Canonical defaults |
|---|---|---|
| ImagCircle | CircleStyle | `wireframe=True`, `wireframe_dash=DottedWireframe()`, `opacity=0.0` |
| ImagSphere | SphereStyle | `wireframe=True`, `wireframe_dash=DottedWireframe()`, `wireframe_opacity=0.6`, `opacity=0.3` |
| ImagPointPair | PointPairStyle | `wireframe=True`, `wireframe_dash=DottedWireframe()`, `wireframe_opacity=0.6`, `color=#ff88ff` |


```python
viz = Visualizer(
    title="Entities - imaginary",
    add_default_axes=False,
    annotation=_NOTE,
)
viz.add(
    ImagCircle(center=Point(0, 0, 0), radius=2, normal=Direction(0, 0, 1)),
)
viz.add(ImagSphere(center=Point(3, 0, 0), radius=1.5))
viz.add(
    ImagPointPair(point_a=Point(-3, -1, 0), point_b=Point(-3, 1, 0)),
)
viz.display_snapshot(height=340)
```
